# PROBABILIDADE FREQUENTISTA

Para encontrar a probabilidade de um determinado evento ocorrer, podemos utilizar a **probabilidade frequentista**. Sendo $A$ um evento aleatório qualquer, podemos encontrar a probabilidade de $A$ utilizando a probabilidade frequentista da seguinte forma:

$$
\hat{P}(A) = \frac{\text{Número de vezes que o evento } A \text{ ocorreu}}{\text{Número total de observações}}
$$

Essa razão representa a **frequência relativa observada** do evento em um número finito de repetições. Na definição frequentista formal, a probabilidade verdadeira $P(A)$ é o **limite dessa frequência relativa** quando o número de observações tende ao infinito.

De tal forma que:

$$
0 \leq P(A) \leq 1
$$

A probabilidade de um evento é sempre um valor entre 0 e 1 (ou em percentual, 0% a 100%).

## Exemplo

Lançamos uma moeda 20 vezes e o evento $A$ = "deu cara" ocorreu 11 vezes. A função `probablidade_frequentista`, definida em `probabilidade.py`, implementa a fórmula acima.

```python
from probabilidade import probablidade_frequentista

freq_evento_a = 11  # número de caras observadas
total_n = 20  # número total de lançamentos

p_hat = probablidade_frequentista(freq_evento_a, total_n)
print(f"P̂(A) = {freq_evento_a}/{total_n} = {p_hat:.2f} ({p_hat:.0%})")
```

## Regra Aditiva

### Exemplo 1 — eventos exclusivos

Uma loja faz uma promoção relâmpago com dois sabores de suco, e cada cliente só pode escolher um: laranja ($P = 0{,}25$) ou uva ($P = 0{,}15$).

### Exemplo 2 — eventos não exclusivos

Numa padaria, $P(\text{pão}) = 0{,}50$, $P(\text{leite}) = 0{,}35$, e a probabilidade de o cliente levar os dois juntos é $P(\text{pão} \cap \text{leite}) = 0{,}20$.

```python
from probabilidade import regra_aditiva

# Exemplo 1 — exclusivos (interseção = 0, valor padrão da função)
p_laranja, p_uva = 0.25, 0.15
p_promocao = regra_aditiva(p_laranja, p_uva)
print(f"P(laranja ∪ uva) = {p_promocao:.2f} ({p_promocao:.0%})")

# Exemplo 2 — não exclusivos
p_pao, p_leite, p_pao_e_leite = 0.50, 0.35, 0.20
p_pao_ou_leite = regra_aditiva(p_pao, p_leite, p_pao_e_leite)
print(f"P(pão ∪ leite) = {p_pao_ou_leite:.2f} ({p_pao_ou_leite:.0%})")
```

## Regra Multiplicativa

Calcula a probabilidade de **dois eventos independentes ocorrerem juntos** (um E outro). A ocorrência de um não afeta a probabilidade do outro:

$$
P(A \cap B) = P(A) \times P(B)
$$

**Exemplo:** numa livraria, a compra de uma revista ($P = 0{,}20$) e a compra de um chocolate no caixa ($P = 0{,}40$) são eventos independentes.

```python
from probabilidade import regra_multiplicativa

p_revista, p_chocolate = 0.20, 0.40
p_ambos = regra_multiplicativa(p_revista, p_chocolate)
print(f"P(revista ∩ chocolate) = {p_ambos:.2f} ({p_ambos:.0%})")
```

## Probabilidade condicional

Calcula a probabilidade de um evento ocorrer **dado que outro já ocorreu**. Lê-se $P(A \mid B)$ como "probabilidade de A dado B":

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}, \quad \text{para } P(B) > 0
$$

**Exemplo:** numa loja de artigos esportivos, $P(\text{tênis}) = 0{,}40$ e $P(\text{tênis} \cap \text{meia}) = 0{,}18$. Qual a chance de o cliente levar meia, dado que já levou o tênis?

```python
from probabilidade import probabilidade_condicional

p_tenis, p_tenis_e_meia = 0.40, 0.18
p_meia_dado_tenis = probabilidade_condicional(p_tenis_e_meia, p_tenis)
print(f"P(meia | tênis) = {p_meia_dado_tenis:.2f} ({p_meia_dado_tenis:.0%})")
```
